```python
#a) Load the iris.csv dataset using Pandas and display the first five records of the dataset.
import pandas as pd
df = pd.read_csv("iris.csv")
print(df.head())
```

```python
#b) Create a scatter plot showing the relationship between sepal.length and sepal.width.
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv("iris.csv")
plt.figure(figsize=(7, 5))
plt.scatter(df["sepal_length"], df["sepal_width"], color="blue")
plt.xlabel("Sepal Length (cm)")
plt.ylabel("Sepal Width (cm)")
plt.title("Sepal Length vs Sepal Width")
plt.grid(True, alpha=0.3)
plt.show()
```

```python
#c) Create a scatter plot of the Iris dataset using different colors for each flower variety (Setosa,Versicolor, and Virginica).
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv("iris.csv")
species_colors = {
    "setosa": "red",
    "versicolor": "green",
    "virginica": "blue"
}
plt.figure(figsize=(7, 5))
for species, color in species_colors.items():
    subset = df[df["species"] == species]
    plt.scatter(subset["sepal_length"], subset["sepal_width"],
                color=color, label=species)
plt.xlabel("Sepal Length (cm)")
plt.ylabel("Sepal Width (cm)")
plt.title("Sepal Length vs Sepal Width by Species")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

```python
#d) Create a line plot showing the relationship between sepal.length and sepal.width.
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv("iris.csv")
df_sorted = df.sort_values("sepal_length")
plt.figure(figsize=(7, 5))
plt.plot(df_sorted["sepal_length"], df_sorted["sepal_width"], color="teal")
plt.xlabel("Sepal Length (cm)")
plt.ylabel("Sepal Width (cm)")
plt.title("Sepal Length vs Sepal Width (Line Plot)")
plt.grid(True, alpha=0.3)
plt.show()
```

```python
#e) Create a histogram showing the distribution of sepal.length values.
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("iris.csv")

plt.hist(df["sepal_length"], bins=10)

plt.xlabel("Sepal Length")
plt.ylabel("Frequency")
plt.title("Distribution of Sepal Length")

plt.show()
```

```python
#f) Create a bar chart showing the total sepal.length for each Iris flower variety.
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("iris.csv")

total_sepal = df.groupby("species")["sepal_length"].sum()

plt.bar(total_sepal.index, total_sepal.values)

plt.xlabel("Iris Variety")
plt.ylabel("Total Sepal Length")
plt.title("Total Sepal Length by Iris Variety")

plt.show()
```

```python
#g) Create a box plot to visualize the distribution of sepal.length for different Iris flower varieties.
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("iris.csv")

df.boxplot(column="sepal_length", by="species")

plt.xlabel("Iris Variety")
plt.ylabel("Sepal Length")
plt.title("Sepal Length Distribution by Iris Variety")
plt.suptitle("")
plt.show()
```

```python
#h) Create a pie chart showing the percentage distribution of the three Iris flower varieties in thedataset.

import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("iris.csv")

species_count = df["species"].value_counts()

plt.pie(
    species_count.values,
    labels=species_count.index,
    autopct="%1.1f%%"
)

plt.title("Percentage Distribution of Iris Varieties")

plt.show()
```

```python
#i) Create a scatter plot comparing petal.length and petal.width, using different colors for each Irisvariety.
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("iris.csv")

setosa = df[df["species"] == "setosa"]
versicolor = df[df["species"] == "versicolor"]
virginica = df[df["species"] == "virginica"]
plt.scatter(
    setosa["petal_length"],
    setosa["petal_width"],
    color="red",
    label="Setosa"
)
plt.scatter(
    versicolor["petal_length"],
    versicolor["petal_width"],
    color="green",
    label="Versicolor"
)
plt.scatter(
    virginica["petal_length"],
    virginica["petal_width"],
    color="blue",
    label="Virginica"
)
plt.xlabel("Petal Length")
plt.ylabel("Petal Width")
plt.title("Petal Length vs Petal Width")
plt.legend()

plt.show()
```

```python
#j
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("iris.csv")

plt.figure(figsize=(10, 8))

plt.subplot(2, 2, 1)
plt.hist(df["sepal_length"], bins=10)
plt.title("Sepal Length")
plt.xlabel("Length")
plt.ylabel("Frequency")

plt.subplot(2, 2, 2)
plt.hist(df["sepal_width"], bins=10)
plt.title("Sepal Width")
plt.xlabel("Width")
plt.ylabel("Frequency")

plt.subplot(2, 2, 3)
plt.hist(df["petal_length"], bins=10)
plt.title("Petal Length")
plt.xlabel("Length")
plt.ylabel("Frequency")

plt.subplot(2, 2, 4)
plt.hist(df["petal_width"], bins=10)
plt.title("Petal Width")
plt.xlabel("Width")
plt.ylabel("Frequency")

plt.tight_layout()
plt.show()
```

```python
#k
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("iris.csv")
average = df.groupby("species")[["sepal_length", "sepal_width"]].mean()

x = range(len(average))
width = 0.35

plt.bar(
    [i - width/2 for i in x],
    average["sepal_length"],
    width=width,
    label="Average Sepal Length"
)

plt.bar(
    [i + width/2 for i in x],
    average["sepal_width"],
    width=width,
    label="Average Sepal Width"
)

plt.xticks(x, average.index)

plt.xlabel("Iris Variety")
plt.ylabel("Average Value")
plt.title("Average Sepal Length and Sepal Width")
plt.legend()

plt.show()
```

```python
#l
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("iris.csv")

bubble_size = df["sepal_length"] * 50

plt.scatter(
    df["petal_length"],
    df["petal_width"],
    s=bubble_size,
    alpha=0.5
)

plt.xlabel("Petal Length")
plt.ylabel("Petal Width")
plt.title("Bubble Scatter Plot")

plt.show()
```
