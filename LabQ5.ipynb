```python
import pandas as pd
import numpy as np
```

```python
# 1) Create a Pandas Series containing the values 1, 3, 5, 6, and 8, and display it.
s = pd.Series([1, 3, 5, 6, 8])
print("1. Pandas Series:")
print(s)
```

```python
# 2) Create a DataFrame from a dictionary containing the details of Country, Capital, Area, and
#Population for the BRICS countries, and set the row labels as BR, RU, IN, CH, and SA.
data = {
    "Country": ["Brazil", "Russia", "India", "China", "South Africa"],
    "Capital": ["Brasilia", "Moscow", "New Delhi", "Beijing", "Pretoria"],
    "Area": [8515767, 17098242, 3287263, 9596961, 1221037],
    "Population": [
        214000000,
        144000000,
        1400000000,
        1410000000,
        60000000
    ]
}

brics = pd.DataFrame(
    data,
    index=["BR", "RU", "IN", "CH", "SA"]
)

print("\n2. BRICS DataFrame:")
print(brics)
```

```python
# 3) Read the dataset cars1.csv into a DataFrame and:
cars = pd.read_csv("cars1.csv")
print("\n3(a). Complete Cars Dataset:")
print(cars)
print("\n3(b). First Four Records:")
print(cars.head(4))
print("\n3(c). Fifth and Sixth Records:")
print(cars.iloc[4:6])
print("\n3(d). Third Row:")
print(cars.iloc[2])
```

```python
# 4) Create a DataFrame containing the columns Name, Gender, and Age, and perform
#DataFrame slicing to:
data = {
    "Name": ["Anu", "Binu", "Cathy", "David", "Esha"],
    "Gender": ["F", "M", "F", "M", "F"],
    "Age": [20, 21, 22, 20, 23]
}

students = pd.DataFrame(data)
print("\n4. Student DataFrame:")
print(students)
print("\nFirst Two Rows:")
print(students.iloc[:2])
print("\nThird Row Onwards:")
print(students.iloc[2:])
```

```python
# 5) Create a Pandas Series containing four random numbers, and display:
random_series = pd.Series(np.random.rand(4))
print("\n5(a). Complete Random Series:")
print(random_series)
print("\n5(b). Values:")
print(random_series.values)
print("\n5(c). First Two Elements:")
print(random_series.head(2))
print("\n5(d). Last Three Elements:")
print(random_series.tail(3))
```

```python
# 6) Create a DataFrame containing the columns Name, Age, and Rating, and display the
#following DataFrame properties:
data = {
    "Name": ["Anu", "Binu", "Cathy", "David"],
    "Age": [20, 21, 22, 23],
    "Rating": [4.5, 4.0, 4.8, 4.2]
}
df = pd.DataFrame(data)
print("\n6. DataFrame:")
print(df)
print("\n(a) Transpose:")
print(df.T)
print("\n(b) Axes:")
print(df.axes)
print("\n(c) Data Types:")
print(df.dtypes)
print("\n(d) Is DataFrame Empty?")
print(df.empty)
print("\n(e) Number of Dimensions:")
print(df.ndim)
print("\n(f) Shape:")
print(df.shape)
print("\n(g) Size:")
print(df.size)
print("\n(h) Values:")
print(df.values)
print("\n(i) Missing Values:")
print(df.isnull().sum())
```
